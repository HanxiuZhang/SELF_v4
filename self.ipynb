```python
import time


ev_eff= "data/ev_eff.csv"
sv_eff= "data/sv_eff.csv"
file_list = [sv_eff, ev_eff]

model_path_qwen=[base_model_qwen]
model_path_qwen.extend(finetune_model_list_qwen + unrelated_model_list_qwen)

start_time =time.perf_counter()

fp_map={}
for item in model_path_qwen:
    fp_map[item] = read_fp(item, file_list, k, m).to(device)

time1 =time.perf_counter()
print(f"finish fp_map:{(time1-start_time):.6f}")


v1 = fp_map[base_model_qwen]
print('--------------Qwen2.5-7B finetune--------------')
for model_path in finetune_model_list_qwen:
    v2 = fp_map[model_path]
    dist = torch.norm(v1-v2, p=1).item()
    print(f"{model_path:60} {dist:>10.8f}")
print('--------------Qwen2.5-7B unrelated--------------')
for model_path in unrelated_model_list_qwen:
    v2 = fp_map[model_path]
    dist = torch.norm(v1-v2, p=1).item()
    print(f"{model_path:60} {dist:>10.8f}")
```

```python
# Llama2 evaluation

model_path_llama2=[base_model_llama2]
model_path_llama2.extend(finetune_model_list_llama2 + prune_model_list_llama2 + unrelated_model_list_llama2)

start_time =time.perf_counter()
fp_map={}
for item in model_path_llama2:
    fp_map[item] = read_fp(item, file_list, k, m).to(device)

time1 =time.perf_counter()
print(f"finish fp_map:{(time1-start_time):.6f}")

v1 = fp_map[base_model_llama2]
print('--------------base_model_llama2 finetune--------------')
for model_path in finetune_model_list_llama2:
    v2 = fp_map[model_path]
    dist = torch.norm(v1-v2, p=1).item()
    print(f"{model_path:60} {dist:>10.8f}")

print('--------------base_model_llama2 pruned--------------')
for model_path in prune_model_list_llama2:
    v2 = fp_map[model_path]
    dist = torch.norm(v1-v2, p=1).item()
    print(f"{model_path:60} {dist:>10.8f}")
    
print('--------------base_model_llama2 unrelated--------------')
for model_path in unrelated_model_list_llama2:
    v2 = fp_map[model_path]
    dist = torch.norm(v1-v2, p=1).item()
    print(f"{model_path:60} {dist:>10.8f}")
```
